Considérez un dé $D_6$ dont les valeurs possibles sont : $\{1, 2, 3, 4, 5, 6\}$.

Ce dé est non truqué, ce qui signifie que toutes les valeurs ont des probabilités égales de se produire.  

Si je jette le dé $D_6$, la probabilité que le résultat soit $3$ est de $\frac{1}{6}$.

1. Quelle est la probabilité d'obtenir $7$ au jet du dé $D_6$ ?
2. Quelle est la probabilité d'obtenir un résultat pair ?
3. Quelle est la probabilité d'obtenir un résultat impair ?
4. Quelle est la probabilité d'obtenir un résultat inférieur ou égal à $4$ ?
5. Quelle est la probabilité d'obtenir un résultat inférieur ou égal à $6$ ?
6. Quelle est la probabilité de ne pas obtenir $1$ ?

1.L'univers des possibles est $\Omega=\{1,2,3,4,5,6\}$.

L'événement « obtenir un 7 » est impossible, car $7\notin\Omega$.

On a donc :

$$A=\{7\}\cap\Omega=\varnothing$$

D'après les axiomes de probabilité, la probabilité de l'ensemble vide est nulle :

$${P(A)=P(\varnothing)=0}$$

-------------------------------------------------------------

2. Si $X = \{2, 4 ,6\} $, par l'axiome d´additivité dénombrable:
$$P(X) = P(\{2\}) + P(\{4\}) +P(\{6\}) = \frac{1}{6} + \frac{1}{6} + \frac{1}{6} = \frac{1}{2}$$


3. Si $X = \{1, 3 ,5\} $, par l'axiome d´additivité dénombrable:
$$P(X) = P(\{1\}) + P(\{3\}) +P(\{5\}) = \frac{1}{6} + \frac{1}{6} + \frac{1}{6} = \frac{1}{2}$$


4. Si $X = \{1, 2 ,3, 4\} $, par l'axiome d´additivité dénombrable:
$$P(X) = P(\{1\}) + P(\{2\}) +P(\{3\})+P(\{4\})= \frac{1}{6} + \frac{1}{6} + \frac{1}{6}+ \frac{1}{6} = \frac{2}{3}$$


5. Si $X = \{1, 2 ,3, 4, 5, 6\} \Rightarrow X = \Omega $, par l'axiome de normalisation:
$$P(\Omega) = 1 $$


6. Si $X= \{1\}$ donc $X^{C}= \{2, 3, 4, 5, 6\}$, comme $X\cup X^{C}=\Omega$, donc $P(X\cup X^{C})=P(X) + P(X^{C})=P(\Omega)$.

```python
import sympy as sp

x = sp.symbols('x')
```

```python
f = 3*x**2 - 2*x + 5
g = 2/x**2
h = sp.sqrt(3*x - 1)
k = sp.exp(x)
l = sp.log(x)
p = sp.sqrt(x) + 2*x**3
q = sp.sin(x)
r = (x**2 + 1)/(x - 1)
```

```python
functions = {
    "f": f,
    "g": g,
    "h": h,
    "k": k,
    "l": l,
    "p": p,
    "q": q,
    "r": r
}

for name, expression in functions.items():
    derivative = sp.simplify(sp.diff(expression, x))
    print(f"{name}'(x) = {derivative}")
```

```python
from IPython.display import display, Math

for name, expression in functions.items():
    derivative = sp.simplify(sp.diff(expression, x))

    display(Math(
        rf"{name}'(x) = {sp.latex(derivative)}"
    ))
```
